## Requirements (from the spec)

1. **Pattern A (handoffs):** triage → billing / technical / account agents, each with its own tools and system prompt.
2. **Pattern B (subagents):** one coordinator calling the same specialists as tools.
3. **Shared RAG** over the product FAQ (`mastery.rag`); **HITL** before any refund over the threshold ($50).
4. **A 15-conversation test set**; compare A vs B **and the single-agent baseline** on resolution correctness, model calls, tokens, $ per conversation and latency (p50/p95). LIVE runs are repeated (`TRIALS`) and graded on the **worst** trial.

**Acceptance criteria:** both patterns pass ≥ 80% of the conversations in every trial (12/15 offline; 7/8 with the LIVE default) · refunds over the threshold always interrupt · the baseline is measured alongside · a written comparison (ADR, status *proposed*) with **your** recommendation.

**Modes:** with `GROQ_API_KEY` the notebook is **LIVE** (`gpt-oss-20b` for every agent; the FAQ is a Chroma index with local embeddings when `fastembed` is installed). Without it, **OFFLINE DEMO**: `mastery.agents.support.offline_model()` (rule-based, reads its role from the system prompt and its options from the tool names) and BM25 over the FAQ. The billing manager is a function (`manager_review`), so Restart & Run All always completes.

````python
def table(rows, columns):
    """Print dict rows as an aligned text table (one comparison cell per Part)."""
    widths = {c: max(len(c), *(len(str(r.get(c, ""))) for r in rows)) for c in columns}
    print("  ".join(c.ljust(widths[c]) for c in columns))
    for r in rows:
        print("  ".join(str(r.get(c, "")).ljust(widths[c]) for c in columns))
````

## Step 6 · Compare A vs B vs the baseline on the test set

````python
# Run every arm on the test set, TRIALS times (fresh graph per trial), conversation by conversation.
def build_baseline(model_for, tools, checkpointer):
    # The single-agent baseline (09.01): one agent, every tool. Multi-agent has to beat THIS.
    return build_single_agent_baseline(model_for("generalist"), tools, checkpointer=checkpointer)


def percentile(values, q):
    values = sorted(values)
    return values[min(len(values) - 1, max(0, math.ceil(q * len(values)) - 1))] if values else 0.0


def run_trial(app):
    rows = []
    for conv in RUN:
        meter = UsageMeter()
        report = evaluate(app, [conv], desk, meter=meter, reviewer=manager_review)
        row = report["rows"][0]
        usage = {"input_tokens": meter.input_tokens, "output_tokens": meter.output_tokens}
        rows.append({**{k: row[k] for k in ("id", "passed", "refund_policy")},
                     "interrupts": row["outcome"]["interrupts"], "answer": row["outcome"]["answer"],
                     "calls": meter.model_calls, "tokens": meter.total_tokens, "usd": estimate_cost(usage, PRICE_MODEL) or 0.0,
                     "seconds": round(meter.seconds, 2), "agents": meter.agents(), "outcome": row["outcome"]})
    return rows


def run_pattern(build):
    if build(model_for, tools, InMemorySaver()) is None:
        return None
    trials = [run_trial(build(model_for, tools, InMemorySaver())) for _ in range(TRIALS)]
    worst = min(trials, key=lambda rows: sum(r["passed"] for r in rows))   # the checks grade the WORST trial
    every = [r for rows in trials for r in rows]
    passed = [sum(r["passed"] for r in rows) for rows in trials]
    return {"rows": worst, "passed": min(passed), "mean_passed": round(statistics.mean(passed), 1), "trials": passed,
            "total": len(worst), "calls": sum(r["calls"] for r in worst), "tokens": sum(r["tokens"] for r in worst),
            "seconds": round(sum(r["seconds"] for r in worst), 2),
            "usd_per_conv": sum(r["usd"] for r in every) / len(every),
            "p50": percentile([r["seconds"] for r in every], 0.50), "p95": percentile([r["seconds"] for r in every], 0.95)}


reports = {"A (handoffs)": run_pattern(build_pattern_a), "B (subagents)": run_pattern(build_pattern_b),
           "baseline (single agent)": run_pattern(build_baseline)}
COLUMNS = ["pattern", "passed (min)", "mean", "model calls", "tokens", "$/conv", "p50 s", "p95 s"]
summary = [{"pattern": name, "passed (min)": f"{r['passed']}/{r['total']}", "mean": r["mean_passed"],
            "model calls": r["calls"], "tokens": f"{r['tokens']:,}", "$/conv": f"{r['usd_per_conv']:.5f}",
            "p50 s": r["p50"], "p95 s": r["p95"]}
           for name, r in reports.items() if r is not None]
print(f"{len(RUN)} conversations x {TRIALS} trial(s) per arm; calls/tokens are the worst trial's")
if reports["A (handoffs)"] or reports["B (subagents)"]:
    table(summary, COLUMNS)
else:
    print("❌ no pattern built yet (TODO 1 and 2)")

if reports["A (handoffs)"] and reports["B (subagents)"]:
    print()
    per_conv = [{"id": a["id"], "A": "✅" if a["passed"] else "❌", "B": "✅" if b["passed"] else "❌",
                 "base": "✅" if c["passed"] else "❌", "A calls": a["calls"], "B calls": b["calls"],
                 "base calls": c["calls"], "A pauses": a["interrupts"], "B pauses": b["interrupts"]}
                for a, b, c in zip(reports["A (handoffs)"]["rows"], reports["B (subagents)"]["rows"],
                                   reports["baseline (single agent)"]["rows"])]
    table(per_conv, ["id", "A", "B", "base", "A calls", "B calls", "base calls", "A pauses", "B pauses"])

# A crashed conversation (pauses = -1) is a provider error, not a wrong answer: read it before you blame your graph.
# "429 ... tokens per day (TPD)" = the free tier's daily budget is spent: rerun tomorrow, or set LIVE_IDS to fewer ids.
errors = [(name, row["id"], row["outcome"]["answer"]) for name, r in reports.items() if r
          for row in r["rows"] if row["outcome"].get("error")]
if errors:
    print(f"\n⚠️ {len(errors)} conversation(s) crashed instead of answering (the worst trial):")
    for name, cid, message in errors:
        print(f"  {name:<24} {cid}  {message}")

saved = {name: None if r is None else {**{k: v for k, v in r.items() if k != "rows"},
                                       "rows": [{k: v for k, v in row.items() if k != "outcome"} for row in r["rows"]]}
         for name, r in reports.items()}
(OUT_DIR / "project09-report.json").write_text(json.dumps(saved, indent=2, default=str), encoding="utf-8")
print("\nsaved data/outputs/m09/project09-report.json")
````

## Step 8 · Acceptance checks

````python
# Acceptance checks (from the spec). ✅ / ❌ per check. Each one fails on the untouched starter.
import re

A, B = reports["A (handoffs)"], reports["B (subagents)"]
BASE = reports["baseline (single agent)"]
REPORTS = [r for r in (A, B) if r is not None]
HITL_IDS = [c["id"] for c in RUN if c["expect"].get("interrupts", 0) > 0]      # S02, S03


def uses(fn, *names):
    # Global names a function's code refers to: did YOU assemble the pattern from the building blocks?
    return set(names) <= set(fn.__code__.co_names)


def rows(rep, conv_id):
    return [r for r in rep["rows"] if r["id"] == conv_id]


def refunds(rep, conv_id):
    return [(f["invoice_id"], float(f["amount"])) for r in rows(rep, conv_id) for f in r["outcome"]["refunds"]]


def adr_section(name):
    found = re.search(rf"^##\s+{name}\b(.*?)(?=^##\s|\Z)", adr_text, flags=re.M | re.S)
    return found.group(1) if found else ""


def evidence_row(rep, *labels):
    # One Evidence table row for this pattern that carries its own measured pass count and model calls.
    return any(line.lstrip().startswith("|") and any(l in line for l in labels)
               and f"{rep['passed']}/{rep['total']}" in line and re.search(rf"\b{rep['calls']}\b", line)
               for line in adr_section("Evidence").splitlines())


checks = {
    "Pattern A is your handoffs graph (triage + 3 specialists, built with handoff_tool + handoff_graph)":
        pattern_a is not None and {"triage", "billing", "technical", "account"} <= set(pattern_a.get_graph().nodes)
        and uses(build_pattern_a, "handoff_tool", "handoff_graph") and not uses(build_pattern_a, "build_handoffs_desk"),
    "Pattern B is your coordinator calling the specialists (built with subagent_tool)":
        B is not None and "coordinator" in B["rows"][0]["agents"] and len(B["rows"][0]["agents"]) > 1
        and uses(build_pattern_b, "subagent_tool") and not uses(build_pattern_b, "build_subagents_desk"),
    f"Pattern A passes >= {THRESHOLD_PASS}/{len(RUN)} in every trial": A is not None and A["passed"] >= THRESHOLD_PASS,
    f"Pattern B passes >= {THRESHOLD_PASS}/{len(RUN)} in every trial": B is not None and B["passed"] >= THRESHOLD_PASS,
    "Refunds over the threshold pause exactly once (S02 and S03, A and B)":
        len(REPORTS) == 2 and {"S02", "S03"} <= set(HITL_IDS) and all(
            r["outcome"]["interrupts"] == 1 for rep in REPORTS for i in HITL_IDS for r in rows(rep, i)),
    "No over-threshold refund issued without a manager (A and B)":
        len(REPORTS) == 2 and all(rep["rows"] for rep in REPORTS)
        and all(r["refund_policy"] for rep in REPORTS for r in rep["rows"]),
    "Manager policy: S02 refunds $199, S03 refunds nothing (A and B), and every policy probe holds":
        POLICY_OK and len(REPORTS) == 2 and all(
            refunds(rep, "S02") == [("INV-2001", 199.0)] and rows(rep, "S03") and refunds(rep, "S03") == []
            for rep in REPORTS),
    "ADR has Context, Evidence, Recommendation and Consequences sections, and names the recommended pattern":
        adr_path.exists() and bool(adr_text) and all(adr_section(h).strip() for h in
                                                    ("Context", "Evidence", "Recommendation", "Consequences"))
        and bool(re.search(r"\b(A|B|handoffs|subagents)\b", adr_section("Recommendation"))),
    "ADR Evidence table: one row per pattern with ITS measured pass count and model calls":
        bool(adr_text) and len(REPORTS) == 2 and evidence_row(A, "A", "handoffs") and evidence_row(B, "B", "subagents"),
    "The single-agent baseline is the third arm: measured on the same conversations and in the Evidence table":
        bool(adr_text) and len(REPORTS) == 2 and BASE["total"] == len(RUN) and evidence_row(BASE, "baseline", "single"),
}
for name, ok in checks.items():
    print("✅" if ok else "❌", name)
print(f"\n{sum(checks.values())}/{len(checks)} acceptance checks passed")
_ = (OUT_DIR / "project09-acceptance.json").write_text(json.dumps(checks, indent=2), encoding="utf-8")
````
